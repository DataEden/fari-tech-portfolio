```python
def interpret_skewness(skew_value) -> str:
    """
    Returns a label describing the skewness direction and strength.
    
    Args:
        skew_value (float): The skewness of the distribution.

    Returns:
        str: A label indicating the skew type.
    """
    if skew_value < -1:
        return "Strong Left Skew"
    elif -1 <= skew_value < -0.5:
        return "Moderate Left Skew"
    elif -0.5 <= skew_value <= 0.5:
        return "Approximately Symmetric"
    elif 0.5 < skew_value <= 1:
        return "Moderate Right Skew"
    else:
        return "Strong Right Skew"
```

```python
# calculate specific skewness
skewness = weather_merged['60kgs_bag'].skew()

# Call interpret_skewness() function
skew_label = interpret_skewness(1.35)
print(f"{skewness:.2f} is {skew_label}")
```

```python
# Calculate skewness 
column = '60kgs_bag'
skew_value = skew(weather_merged[column])
label = interpret_skewness(skew_value)

print(f"Column: '{column}' → Skewness: {skew_value:.2f} → {label}")
```

```python
x = weather_merged['temp_avg']
y = weather_merged['60kgs_bag']

r_value, p_value = pearsonr(x, y)

plt.figure(figsize=(8,6))
sns.regplot(x=x, y=y, ci=95, line_kws={"color": "black"})
plt.title("Average Temperature vs Coffee Yield")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Coffee Production (Million 60kg Bags)")

plt.annotate(f"r = {r_value:.2f}\np = {p_value:.2f}",
             xy=(0.05, 0.90),
             xycoords='axes fraction')

plt.tight_layout()
plt.show()
```

## Writeup

**Relationship Between temp_avg and 60kgs_bag**

The Pearson correlation coefficient between average temperature and annual coffee production was r ≈ -0.04 (p = 0.71), indicating no statistically significant linear relationship at the annual aggregation level.

This suggests that moderate variation in average growing-season temperature does not independently explain fluctuations in total harvest output. The lack of significance implies that temperature effects may either be non-linear, buffered by agricultural management practices, or overshadowed by structural production factors such as cultivated land area and bearing tree density.

**Relationship Between hum_min and 60kgs_bag**

The Pearson correlation coefficient between minimum humidity and coffee production was r ≈ -0.06 (p ≈ 0.55), indicating a very weak and statistically insignificant relationship.

Although fungal threats such as coffee leaf rust are commonly associated with humid conditions, this analysis suggests that annual minimum humidity levels alone do not materially influence total production volume. This may indicate that disease risk depends more heavily on sustained humidity patterns, localized microclimates, or multi-variable interactions rather than isolated annual minimum readings.

**Relationship Between rain_max and 60kgs_bag**

The correlation between maximum rainfall and coffee output was r ≈ -0.02 (p ≈ 0.82), demonstrating virtually no measurable linear association.

This indicates that peak rainfall events during the growing season do not significantly predict annual production levels in this dataset. It is plausible that rainfall effects operate through cumulative seasonal totals, timing-specific distribution, or soil absorption dynamics rather than single extreme measurements.

**Relationship Between wind_avg and 60kgs_bag**

The Pearson correlation coefficient between average wind speed and coffee production was r ≈ -0.13 (p ≈ 0.21), indicating a weak negative relationship that is not statistically significant.

While wind can influence pollination, disease spread, and plant stress, the data suggests that average seasonal wind speeds are not a primary linear driver of annual harvest volume. The influence of wind may instead depend on episodic extreme events or interact with other climatic variables.

**Weather Variables vs. Structural Coffee Variables**

When examining the relationship between key weather variables (temp_avg, hum_min, rain_max, wind_avg) and structural production metrics such as bearing_trees or bearing_hectares, correlations remained weak and statistically insignificant (e.g., r ≈ -0.08, p > 0.05).

This suggests that annual weather averages may not be strong standalone predictors of structural agricultural capacity. Instead, production capacity appears more strongly influenced by long-term land utilization, cultivation strategy, and tree density management.

Which Weather Factor Influences Coffee Production the Most?

Across all evaluated weather variables, no single factor demonstrated a strong or statistically significant linear relationship with annual coffee yield.

Relative comparisons suggest that structural production variables (e.g., bearing hectares, tree density) exhibit stronger directional relationships with yield than weather aggregates. This implies that production output may be capacity-driven rather than climate-driven at annual resolution.

**Additional Data to Improve Analysis**

To better understand weather-production dynamics, additional data should include:

- Monthly or daily weather granularity rather than annual averages
- Cumulative rainfall during flowering and fruiting phases
- Soil moisture and drought severity indices
- Fertilizer usage and irrigation data
- Pest and disease incidence reports
- Coffee commodity price data (to evaluate economic influence on cultivated area)

More granular temporal resolution would allow lagged correlation modeling and potentially uncover non-linear or threshold effects currently obscured in annual aggregates.

**Most Favorable Weather Conditions for Coffee Production**

Based on the statistical results of this analysis, no single annual weather variable demonstrated a strong linear effect on production.

However, the findings suggest that stable growing-season conditions — moderate temperatures, controlled humidity levels, and absence of extreme rainfall or wind events — may provide an environment conducive to consistent output.

More importantly, structural agricultural capacity (bearing hectares and tree density) appears to exert stronger influence over annual yield than seasonal climate variability at the aggregation level analyzed.

### Analytical Limitation Note

- This analysis relies on annual aggregated weather metrics. 
- Future modeling efforts using higher-frequency temporal data may reveal lagged or non-linear relationships not detectable under yearly averaging.
